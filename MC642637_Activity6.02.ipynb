```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
movies = pd.read_csv('movie_scores.csv')
movies.head()
```

```python
labels = movies['MovieTitle']
tomatometer = movies['Tomatometer']
audience = movies['AudienceScore']

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6), dpi=150)
rects1 = ax.bar(x - width/2, tomatometer, width, label='Tomatometer')
rects2 = ax.bar(x + width/2, audience,    width, label='Audience Score')

ax.set_ylabel('Percentage')
ax.set_title('Movie Scores Comparison')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

ax.set_yticks(np.arange(0, 101, 20))
ax.set_yticks(np.arange(0, 101, 5), minor=True)
ax.yaxis.grid(which='major')
ax.yaxis.grid(which='minor', linestyle='--')
plt.tight_layout()
plt.show()
```
